```python
from bs4 import BeautifulSoup
import re # Import regular expression module

# Clear empty_list to populate it with structured data
empty_list = []

for file_path in files:
    with open(file_path, 'r', encoding='utf-8') as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, 'html.parser')

    # Each article is typically enclosed in a div with an id starting with 'article-'
    # and has class 'article'
    articles = soup.find_all('div', id=re.compile(r'^article-'), class_='article')

    for article_soup in articles:
        headline = article_soup.find('span', class_='enHeadline')
        publication_date = article_soup.find('div', string=re.compile(r'\d{4} 年 \d{1,2} 月 \d{1,2} 日')) # Find div containing date in Chinese format
        source_name = article_soup.find('span', class_='colorLinks')
        author_name = article_soup.find('div', class_='author')

        # Extract all paragraph texts for the body
        body_paragraphs = article_soup.find_all('p', class_='articleParagraph enarticleParagraph')
        body_text = ' '.join([p.get_text(strip=True) for p in body_paragraphs])

        # Extract relevant text, handling potential None values
        article_data = {
            'Headline': headline.get_text(strip=True) if headline else None,
            'Publication_Date_Raw': publication_date.get_text(strip=True) if publication_date else None, # Keep raw for conversion later
            'Source_Name': source_name.get_text(strip=True) if source_name else None,
            'Author_Name': author_name.get_text(strip=True) if author_name else None,
            'CombinedText': body_text
        }
        empty_list.append(article_data)

print(f"Populated empty_list with {len(empty_list)} articles.")
# Display first item in empty_list to verify structure
if empty_list:
    print("First extracted article data:")
    print(empty_list[0])
else:
    print("No articles were extracted. Please check HTML structure and selectors.")
```

```python
import os
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# 如果没有下载过 nltk 停用词，需要先运行一次

import nltk
nltk.download('punkt')
nltk.download('stopwords')

# 文件夹路径

txt_folder = '/content/drive/MyDrive/factiva_energy_project/factiva' # Use the correct path for the created subdirectory

# 自定义停用词（可根据需求添加更多）

custom_stoplist = set([
    'said', 'also', 'would', 'could', 'mr', 'ms', 'new', 'one', 'two',
    'energy', 'policy', 'government', 'minister', 'report', 'years','must','works','action','people','world','commission','national','may','year','work','member','international','first','time'
])

# 加载英语停用词

stop_words = set(stopwords.words('english')).union(custom_stoplist)

# 用于存储所有文章的 token

all_tokens = []

# Ensure the directory exists before listing its contents
os.makedirs(txt_folder, exist_ok=True)

for filename in os.listdir(txt_folder):
    if filename.endswith('.txt'):
        with open(os.path.join(txt_folder, filename), 'r', encoding='utf-8') as f:
            text = f.read()
        # 小写
        text = text.lower()
        # 分词
        tokens = word_tokenize(text)
        # 保留字母单词，长度 >= 3
        tokens = [t for t in tokens if t.isalpha() and len(t) >= 3]
        # 去停用词
        tokens = [t for t in tokens if t not in stop_words]

        all_tokens.extend(tokens)

# 检查前 20 个 token

print(all_tokens[:20])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# convert a collection of text documents (your speeches) into a matrix where:
# Each row = one document (a speech);
# Each column = one term (a word);
# Each cell value = TF–IDF weight of that term in that document
# min_df=2 → ignore words that appear in fewer than 2 documents
tfidf = TfidfVectorizer(min_df=2)

# Feeds your cleaned text (from the text_clean column) into the vectorizer.
# Two steps happen in one command:
# .fit() — learns the vocabulary and IDF (Inverse Document Frequency) weights.
# .transform() — applies the TF–IDF transformation to each document.
# Returns a sparse matrix X_tfidf of shape:
X_tfidf = tfidf.fit_transform(df['text_clean'])

# Retrieves the list of all terms (vocabulary) that the vectorizer kept.
# Converts it into a NumPy array for easy indexing and sorting later.
# You’ll use it when finding the top TF–IDF terms for each speech:
terms = np.array(tfidf.get_feature_names_out())

# ((# of docs, # of unique words), # of unique words)
print(X_tfidf.shape, len(terms))
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Ensure 'Publication_Date' is datetime and 'year' is derived
df['Publication_Date'] = pd.to_datetime(df['Publication_Date'], errors='coerce')
df['year'] = df['Publication_Date'].dt.year

# Ensure 'energy topic' column exists in df
# For now, it will be a placeholder. If actual energy topic extraction is needed, that would be a separate task.
if 'energy topic' not in df.columns:
    df['energy topic'] = 'Unknown Energy Topic' # Initialize if not present

# Pick a speech by index (0=earliest, -1=latest)
doc_idx = len(df) - 1  # last speech by default

# Check if X_tfidf and terms are available from previous steps
if 'X_tfidf' in locals() and 'terms' in locals() and X_tfidf is not None and terms is not None:
    # Helper function to get top TF-IDF terms for a given document
    def top_tfidf_terms_for_doc(doc_idx, top_n=100):
        # Get the TF-IDF vector for the specified document
        doc_vector = X_tfidf[doc_idx, :].toarray().flatten()

        # Get indices of top_n highest TF-IDF scores
        top_indices = doc_vector.argsort()[-top_n:][::-1]

        # Get corresponding terms and scores
        top_terms = [(terms[i], doc_vector[i]) for i in top_indices]
        return top_terms

    # Generate dictionary of top TF–IDF terms
    wc_data = dict(top_tfidf_terms_for_doc(doc_idx, top_n=100))

    # Create and display the word cloud
    wc = WordCloud(width=800, height=400, background_color='white')
    wc.generate_from_frequencies(wc_data)

    plt.figure(figsize=(10,5))
    plt.imshow(wc, interpolation='bilinear')
    plt.axis('off')
    plt.title(f"Word Cloud for {df['year'].iloc[doc_idx]} - {df['energy topic'].iloc[doc_idx]}", fontsize=14) # Dynamic title
    plt.show()
else:
    print("TF-IDF data (X_tfidf or terms) not found. Please ensure previous steps ran correctly.")

```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import numpy as np
import pandas as pd

# --- Step 1: Ensure Publication_Date is datetime ---

df['Publication_Date'] = pd.to_datetime(df['Publication_Date'], errors='coerce')  # invalid dates become NaT

# Extract year safely

df['year'] = df['Publication_Date'].dt.year  # NaT will become NaN

# Optional: fill missing years with placeholder (0 or 'Unknown')

df['year'] = df['year'].fillna(0).astype(int)  # now all years are integers

# --- Step 2: Bag-of-words and LDA ---

cv = CountVectorizer(min_df=2)
X_counts = cv.fit_transform(df['text_clean'])
vocab = np.array(cv.get_feature_names_out())

K = 8  # number of topics
lda = LatentDirichletAllocation(
n_components=K,
learning_method="batch",
random_state=42,
max_iter=20
)

topic_mix = lda.fit_transform(X_counts)  # theta: (n_docs, K)

# --- Step 3: Show top words per topic ---

def show_topics(model, vocab, topn=12):
    for k, comp in enumerate(model.components_):
        top_idx = comp.argsort()[::-1][:topn]
        print(f"\nTopic {k}: " + ", ".join(vocab[top_idx]))

show_topics(lda, vocab, topn=12)

# --- Step 4: Ensure energy topic column exists ---

if 'energy topic' not in df.columns:
    df['energy topic'] = 'Unknown Energy Topic'

# --- Step 5: Assemble per-document topic proportions ---

topic_df = pd.DataFrame(topic_mix, columns=[f"topic_{k}" for k in range(K)])

# Concatenate with Headline, year, and energy topic

result_df = pd.concat([df[['Headline', 'year', 'energy topic']], topic_df], axis=1)

# --- Step 6: Preview ---

result_df.head(5)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import numpy as np # Ensure numpy is imported for array operations
import pandas as pd # Import pandas
import re # Import regular expression module
from nltk.corpus import stopwords # Import stopwords
from nltk.tokenize import word_tokenize # Import word_tokenize
import nltk # Import nltk

# Ensure NLTK data is downloaded if not already
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True) # Added to fix LookupError

# Define custom stop words and load English stop words
custom_stoplist = set([
    'said', 'also', 'would', 'could', 'mr', 'ms', 'new', 'one', 'two',
    'energy', 'policy', 'government', 'minister', 'report', 'years','must','works','action','people','world','commission','national','may','year','work','member','international'
])
stop_words = set(stopwords.words('english')).union(custom_stoplist)

# Re-initialize df from empty_list to ensure it's populated
# Assuming empty_list is correctly populated by Hv5THTowkQMo
if 'empty_list' in globals() and len(empty_list) > 0:
    df = pd.DataFrame(empty_list)
else:
    print("Error: 'empty_list' is not populated. Please run HTML parsing cells first.")
    df = pd.DataFrame()

if not df.empty:
    # Convert 'Publication_Date_Raw' to datetime and extract year
    if 'Publication_Date_Raw' in df.columns:
        df['Publication_Date'] = df['Publication_Date_Raw'].apply(lambda x: pd.to_datetime(x, format='%Y 年 %m 月 %d 日', errors='coerce'))
        df['year'] = df['Publication_Date'].dt.year.fillna(0).astype(int)
    else:
        df['Publication_Date'] = pd.NaT
        df['year'] = 0 # Default year if date parsing fails

    # Sort values by Publication_Date (if it's not NaT)
    df.sort_values(by='Publication_Date', inplace=True)
    df.reset_index(drop=True, inplace=True)

    # Ensure 'energy topic' column exists
    if 'energy topic' not in df.columns:
        df['energy topic'] = 'Unknown Energy Topic' # Initialize if not present

    # Define text cleaning function (re-defined to ensure it's available)
    def clean_text(text):
        if pd.isna(text):
            return ""
        text = str(text).lower()  # Convert to lowercase
        text = re.sub(r'\b\w{1,2}\b', '', text)  # Remove words with 1 or 2 characters
        text = re.sub(r'[^a-z\s]', '', text)  # Remove non-alphabetic characters
        tokens = word_tokenize(text)
        tokens = [word for word in tokens if word.isalpha() and word not in stop_words]
        return ' '.join(tokens)

    # Apply cleaning to 'CombinedText' to create 'text_clean'
    if 'CombinedText' in df.columns:
        df['text_clean'] = df['CombinedText'].apply(clean_text)
    else:
        df['text_clean'] = "" # Fallback if CombinedText is missing

    # Bag-of-words for LDA
    cv = CountVectorizer(min_df=2)
    X_counts = cv.fit_transform(df['text_clean'])
    vocab = np.array(cv.get_feature_names_out())

    # Train LDA
    K = 8  # number of topics

    lda = LatentDirichletAllocation(
        n_components=K,
        learning_method="batch",
        random_state=42,
        max_iter=20
    )

    topic_mix = lda.fit_transform(X_counts)  # theta: (n_docs, K)

    def show_topics(model, vocab, topn=12):
        for k, comp in enumerate(model.components_):
            top_idx = comp.argsort()[::-1][:topn]
            print(f"\nTopic {k}: " + ", ".join(vocab[top_idx]))

    show_topics(lda, vocab, topn=12)

    # Assemble per-document topic proportions
    topic_df = pd.DataFrame(topic_mix, columns=[f"topic_{k}" for k in range(K)])
    # Correcting the column selection to include 'energy topic' from df, and removing 'president'
    result_df = pd.concat([df[['Headline','year','energy topic']], topic_df], axis=1)
    print(result_df.head(5))
else:
    print("DataFrame 'df' is empty. Cannot perform LDA.")
```

```python
def topic_top_words(lda_model, vocab, topn=10):
    """Return:
       - topic_labels: list like ["T0: economy, growth, jobs", ...]
       - topic_words:  list of lists of the topn words per topic (for hover)
    """
    labels = []
    words_list = []
    for k, comp in enumerate(lda_model.components_):
        top_idx = comp.argsort()[::-1][:topn]
        words = vocab[top_idx].tolist()
        words_list.append(words)
        label = f"T{k}: " + ", ".join(words[:6])  # concise label for axis/hover
        labels.append(label)
    return labels, words_list

topic_labels, topic_words = topic_top_words(lda, vocab, topn=12)

# Columns in result_df that are topic proportions
topic_cols = [c for c in result_df.columns if c.startswith("topic_")]
```

```python
import plotly.graph_objects as go
import numpy as np

# 确保 topic proportions 归一化

result_df[topic_cols] = result_df[topic_cols].div(result_df[topic_cols].sum(axis=1), axis=0)

# 构建 Z 矩阵

Z = result_df[topic_cols].to_numpy()

# X 轴：主题编号

x = [f"T{int(c.split('_')[-1])}" for c in topic_cols]

# Y 轴：年份 + energy topic + Headline 前20个字符

y = [f"{r['year']} — {r['energy topic']} ({r['Headline'][:20]}...)" for _, r in result_df.iterrows()]

# 构建 hovertext 矩阵

hovertext = []
for _, r in result_df.iterrows(): # Fix: Changed '*' to '_'
    row_texts = []
    for t_i, col in enumerate(topic_cols):
        k = int(col.split('_')[-1]) # Fix: Changed '*' to '_'
        row_texts.append(
            f"<b>{r['energy topic']}</b><br>"
            f"Topic {k}: {r[col]:.3f}<br>"
            f"Top words: {', '.join(topic_words[k][:10])}"
        )
    hovertext.append(row_texts)

# 动态 zmax

zmax = np.max(Z)

# 创建 heatmap

fig = go.Figure(
    data=go.Heatmap(
        z=Z,
        x=x,
        y=y,
        colorscale="Plasma",
        zmin=0.0, zmax=zmax,
        hoverinfo="text",
        text=hovertext,
        colorbar=dict(title="Topic Proportion")
    )
)

fig.update_layout(
    title="Topic mixture (theta) — All Energy Reporting Articles",
    xaxis_title="Topic",
    yaxis_title="Articles",
    height=600,  # 可根据需要调整高度
    margin=dict(l=200, r=20, t=80, b=60)  # 左侧留足空间显示长标签
)

# 启用交互式缩放

fig.update_yaxes(automargin=True)
fig.show()
```
